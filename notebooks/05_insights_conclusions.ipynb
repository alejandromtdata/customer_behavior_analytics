# 5. Insights y conclusiones

## Objetivo

En esta última parte se resumen los principales resultados obtenidos durante el análisis de los datos.

El objetivo es identificar patrones en el comportamiento de los usuarios, su uso de la plataforma, su retención y su relación con la generación de ingresos.

## Principales aprendizajes

A partir del análisis realizado, podemos observar que los usuarios no tienen todos el mismo comportamiento.

Algunos usuarios utilizan la plataforma de forma más frecuente, mientras que otros tienen una actividad más limitada. Esto hace que sea importante analizar los diferentes comportamientos y no considerar a todos los usuarios como un único grupo.

También se observa que la actividad de los usuarios está relacionada con su permanencia en la plataforma. Los usuarios que mantienen una mayor actividad tienen una mayor probabilidad de seguir utilizando la plataforma durante más tiempo.

## Retención y cohortes

El análisis de cohortes nos permite observar cómo evoluciona la retención de los usuarios después de registrarse.

Los principales resultados son:

- La retención es más alta durante los primeros meses.
- La cantidad de usuarios activos disminuye conforme aumenta el tiempo desde el registro.
- Existen diferencias entre las distintas cohortes, aunque todas siguen en general una tendencia de descenso.
- La pérdida de usuarios durante los primeros meses parece ser uno de los puntos más importantes a tener en cuenta.

Este análisis permite identificar los primeros meses después del registro como una etapa importante para intentar mejorar la retención.

## Ingresos

El análisis de las transacciones también muestra que la generación de ingresos no está repartida de la misma forma entre todos los usuarios.

Algunos usuarios generan un mayor valor económico que otros, por lo que puede ser interesante estudiar qué características tienen estos usuarios y qué comportamiento presentan.

Combinar la información de actividad, retención e ingresos permite tener una visión más completa del comportamiento de los usuarios.

## Recomendaciones

A partir de los resultados obtenidos, algunas posibles líneas de actuación serían:

- Mejorar el proceso de onboarding para facilitar los primeros pasos de los nuevos usuarios.
- Analizar qué ocurre durante los primeros meses, ya que es cuando se produce una mayor pérdida de usuarios.
- Fomentar el uso frecuente de las principales funcionalidades de la plataforma.
- Analizar con más detalle los usuarios que presentan una mayor actividad y generan más ingresos.
- Estudiar posibles estrategias para aumentar la retención de los usuarios con menor actividad.

Estas recomendaciones serían un punto de partida para realizar análisis posteriores y comprobar qué acciones podrían tener un mayor impacto.

## Conclusión final

Este proyecto me ha permitido analizar el comportamiento de los usuarios desde diferentes puntos de vista, utilizando información sobre usuarios, sesiones, transacciones y retención.

A través del análisis se ha podido observar que los usuarios presentan comportamientos diferentes y que la retención disminuye principalmente a medida que aumenta el tiempo desde el registro.

Uno de los puntos más importantes que se pueden extraer del análisis es la importancia de los primeros meses de uso. Entender qué ocurre durante esta etapa podría ayudar a mejorar la retención y conseguir que los usuarios continúen utilizando la plataforma.

En general, el proyecto me ha servido para practicar un proceso completo de análisis de datos: explorar los datos, calcular métricas, analizar el comportamiento de los usuarios, estudiar la retención mediante cohortes y obtener conclusiones a partir de los resultados.
